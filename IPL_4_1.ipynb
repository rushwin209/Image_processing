```python
from PIL import Image
import matplotlib.pyplot as plt
from google.colab import files
uploaded=files.upload()

img=Image.open("lena.jpg")
gray=img.convert("L")
hsv=img.convert("HSV")
cymk=img.convert("CMYK")
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.imshow(img)
plt.title("RGB image")
plt.axis("off")
plt.subplot(2,2,2)
plt.imshow(gray,cmap="gray")
plt.title("Gray image")
plt.axis("off")
plt.subplot(2,2,3)
plt.imshow(hsv)
plt.title("HSV image")
plt.axis("off")
plt.subplot(2,2,4)
plt.imshow(cymk)

plt.title("CMYK image")
plt.axis("off")
plt.show()
```
