```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
uploaded = files.upload()
img=cv2.imread("lena.jpg",0)

noise=img.copy()

salt=np.random.rand(*img.shape)<0.2
noise[salt]=255

pepper=np.random.rand(*img.shape)<0.02
noise[pepper]=0

restored=cv2.medianBlur(noise,11)
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(img,cmap="gray")
plt.title("original")
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(noise,cmap="gray")
plt.title("noisy")
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(restored,cmap="gray")
plt.title("restored")
plt.axis("off")
plt.show()
```
