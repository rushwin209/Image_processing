```python
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
from google.colab import files
import io

img=Image.open("./grayscaleimage.jfif").convert("L")
equalized = ImageOps.equalize(img)

plt.figure(figsize=(10,10))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(equalized, cmap='gray')
plt.title('Equalized')
plt.axis("off")

plt.subplot(2,2,3)
plt.hist(img.getdata(), bins=256, color="blue")
plt.title("Original Histogram")
plt.xlim([0,256])

plt.subplot(2,2,4)
plt.hist(equalized.getdata(), bins=256, color="orange")
plt.title("Equalized Histogram")
plt.xlim([0,256])

plt.show()

equalized.save("Equalized.jpg")
```

```python
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
from google.colab import files
import io

img = Image.open("./colorimage.jfif").convert("RGB")
equalized = ImageOps.equalize(img)

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title('Original')
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(equalized)
plt.title('Equalized')
plt.axis("off")

plt.subplot(2, 2, 3)
colors = ('red', 'green', 'blue')
for i, color in enumerate(colors):
    channel_data = img.getchannel(i).getdata()
    plt.hist(channel_data, bins=256, color=color, alpha=0.5, label=color.upper())
plt.title("Original Histogram")
plt.xlim([0, 256])
plt.legend()

plt.subplot(2, 2, 4)
for i, color in enumerate(colors):
    channel_data = equalized.getchannel(i).getdata()
    plt.hist(channel_data, bins=256, color=color, alpha=0.5, label=color.upper())
plt.title("Equalized Histogram")
plt.xlim([0, 256])
plt.legend()

plt.show()

equalized.save("Equalized_Color.jpg")
```
